### the mean square error loss function

```python
def loss_function(m,b,points):
    total_error = 0 
    for i in range(len(points)):
        x = points.iloc[i,0]
        y = points.iloc[i,1]
        total_error += (y - (m*x + b))**2
    mse = total_error/float(len(points))
```

### the gradient descent function

```python
def gradient_descent(m_now,b_now,points,L):
    m_gradient = 0
    b_gradient = 0

    n = len(points)

    for i in range(n):
        x = points.iloc[i,0]
        y = points.iloc[i,1]

        m_gradient += -(2/n)*(x)*(y-(m_now*x+b_now))
        b_gradient += -(2/n)*(y-(m_now*x+b_now))

    m_now -= L*m_gradient
    b_now -= L*b_gradient

    return m_now,b_now

    
```

### the training part

```python
m = 0
b = 0
L = 0.0001
epochs = 500

for i in range(epochs):
    if i % 50 == 0:
        print(f"The Epoch is : {i}")
    m,b = gradient_descent(m,b,data,L)

print(m,b)
```
